```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('wbc.csv')
print("=== Initial Dataset Info ===")
print(f"Dataset Shape: {df.shape}")
print(df.info())
```

Perform encoding on the "diagnosis" column.



```python
df_cleaned = df.drop(columns=['id', 'Unnamed: 32'])
```

```python
le = LabelEncoder()
df_cleaned['diagnosis'] = le.fit_transform(df_cleaned['diagnosis'])

print("\n=== Task 2: Diagnosis Encoding ===")
for index, class_label in enumerate(le.classes_):
    print(f"Class '{class_label}' -> Encoded Value: {index}")

print("\nValue Counts for Encoded Target ('diagnosis'):")
print(df_cleaned['diagnosis'].value_counts())
```

Perform standardization on all columns that possess numerical values.

```python
numerical_cols = [col for col in df_cleaned.columns if col != 'diagnosis']

scaler = StandardScaler()
df_cleaned[numerical_cols] = scaler.fit_transform(df_cleaned[numerical_cols])

df_cleaned.head()
```
